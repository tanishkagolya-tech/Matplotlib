###Scatter Plots

Bivariate Analysis

numerical vs numerical

Use case - Finding correlation

```python
# plt.scatter simple function
x = np.linspace(-10,10,50)

y = 10*x + 3 + np.random.randint(0,300,50)
y
```

This code creates sample data for a scatter plot. First, np.linspace(-10, 10, 50) generates 50 evenly spaced values between -10 and 10 and stores them in x. Then, the y values are calculated using the linear equation y=10x+3, which creates a positive linear relationship between x and y. To make the data more realistic, random values between 0 and 299 are added to each y value using np.random.randint(0, 300, 50). This random addition is called noise, and it causes the points to be scattered around the line instead of lying exactly on it.

```python
plt.scatter(x,y)
```

Difference between with 'o' and without 'o'

Without 'o'

plt.plot(x, y)

Shows a line graph

Points are connected with a line

Used to show trend



With 'o'

plt.plot(x, y, 'o')

Shows only dots (markers)

No connecting line

Used to show individual data points

Key idea:

No 'o' → line plot

'o' → scatter-like dots only

##Bar chart

Bivariate Analysis

Numerical vs Categorical

Use case - Aggregate analysis of groups

```python
# simple bar chart
children = [10,20,40,10,30]
colors = ['red','blue','green','yellow','pink']

plt.bar(colors,children,color='pink')
```

```python
# horizontal bar chart
plt.barh(colors,children,color='black')
```

```python
# color and label
df = pd.read_csv('batsman_season_record.csv')
df
```

```python
plt.bar(df['batsman'],df['2015'],color='pink')
```

```python
plt.bar(np.arange(df.shape[0]) - 0.2,df['2015'],width=0.2,color='yellow')
plt.bar(np.arange(df.shape[0]),df['2016'],width=0.2,color='red')
plt.bar(np.arange(df.shape[0]) + 0.2,df['2017'],width=0.2,color='blue')

plt.xticks(np.arange(df.shape[0]), df['batsman'])

plt.show()
```

```python
# a problem
children = [10,20,40,10,30]
colors = ['red red red red red red','blue blue blue blue','green green green green green','yellow yellow yellow yellow ','pink pinkpinkpink']

plt.bar(colors,children,color='black')
plt.xticks(rotation='vertical')
```

```python
# Stacked Bar chart
plt.bar(df['batsman'],df['2017'],label='2017')
plt.bar(df['batsman'],df['2016'],bottom=df['2017'],label='2016')
plt.bar(df['batsman'],df['2015'],bottom=(df['2016'] + df['2017']),label='2015')

plt.legend()
plt.show()
```

##Pie chart

Univariate/Bivariate Analysis

Categorical vs numerical

Use case - To find contibution on a standard scale

```python
data = [19,20,40,50,60,89]
subject = ['MATHS','SCIENCE','SST','GK','ENGLISH','HINDI']
```

```python
plt.pie(data,labels=subject)
plt.show()
```

```python
# dataset
df = pd.read_csv('gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%')
plt.show()
```

```python
# percentage and colors
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',colors=['blue','green','yellow','pink','cyan','brown'])
plt.show()
```

```python
# explode shadow
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',explode=[0.3,0,0,0,0,0.1],shadow=True)
plt.show()
```
