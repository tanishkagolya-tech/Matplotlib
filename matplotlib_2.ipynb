````python
fig = plt.figure()

ax1 = fig.add_subplot(2,2,1)
ax1.scatter(batters['avg'],batters['strike_rate'],color='red')

ax2 = fig.add_subplot(2,2,2)
ax2.hist(batters['runs'])

ax3 = fig.add_subplot(2,2,3)
ax3.hist(batters['avg'])
````

This code is creating **multiple plots in the same figure** using **subplots**.
fig = plt.figure()

## 1. Create a Figure
fig = plt.figure()

## 2. First Subplot

ax1 = fig.add_subplot(2,2,1)

### Meaning of (2,2,1`


add_subplot(rows, columns, position)

(2,2,1)

means:

* 2 rows
* 2 columns
* Position 1
+---------+---------+
|    1    |    2    |
+---------+---------+
|    3    |    4    |
+---------+---------+
```

This creates graph area **1**.

ax1.scatter(batters['avg'],batters['strike_rate'],color='red')

Draws a scatter plot.

* X-axis → Average
* Y-axis → Strike Rate
* Color → Red

## 3. Second Subplot


ax2 = fig.add_subplot(2,2,2)


Creates graph area **2**.

```text
+---------+---------+
|    1    |    2 ←  |
+---------+---------+
|    3    |    4    |
+---------+---------+

ax2.hist(batters['runs'])


Creates a histogram of runs.

Shows how runs are distributed.

Example:


Frequency
 ^
 |
 |      ███
 |   ███████
 |██████████
 +---------------> Runs

## 4. Third Subplot


ax3 = fig.add_subplot(2,2,3)


Creates graph area **3**.

+---------+---------+
|    1    |    2    |
+---------+---------+
|  3 ←    |    4    |
+---------+---------+

ax3.hist(batters['avg'])

Creates histogram of batting averages.

Shows:

* How many players have avg between 20-30
* How many have avg between 30-40
* etc.

---

## Final Layout

+----------------+----------------+
| Scatter Plot   | Runs Histogram |
| Avg vs SR      |                |
+----------------+----------------+
| Avg Histogram  | Empty          |
+----------------+----------------+
```

The 4th subplot is empty because you haven't created:


ax4 = fig.add_subplot(2,2,4)

fig.add_subplot(a,b,c)


where:

* `a` = rows
* `b` = columns
* `c` = position number

````python
#3D line plot
x = [0,1,5,25]
y = [0,10,13,0]
z = [0,13,20,9]

fig = plt.figure()

ax = plt.subplot(projection='3d')

ax.scatter3D(x,y,z,s=[100,100,100,100])
ax.plot3D(x,y,z,color='red')
````

This code creates a 3D graph, places 4 points in space using (x,y,z) coordinates, makes them visible with scatter3D(), and then joins them with a red line using plot3D().

````python
#3D Surface Plot
x=np.linspace(-10,10,100)
y=np.linspace(-10,10,100)

````

````python
xx,yy=np.meshgrid(x,y)
````

````python
xx
````

````python
yy
````

````python
z = xx**2 + yy**2
z.shape
````

````python
fig = plt.figure(figsize=(12,8))
ax=plt.subplot(projection='3d')
p=ax.plot_surface(xx,yy,z,cmap='flare')
fig.colorbar(p)
````

````python
z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
````

````python
z = np.sin(xx) + np.log(xx)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)
````
